# Strategy Label Schema

The `strategies.json` file (generated by `quickstart.py -s`) uses the following labeling convention:

| Value | Meaning |
|-------|---------|
| `0` | Not cheese |
| `1` | Cheese |
| `999` | Not a strategy |

### `strategies.json` structure

```json
{
    "PlayerName": {
        "StrategyName": "<ADD_LABEL_HERE>",
        "AnotherStrategy": "1"
    }
}
```

- Each **player name** is a top-level key
- Each **unique strategy** (from `Tag:` messages) is a nested key under that player
- Values are labels: `"<ADD_LABEL_HERE>"` for unlabeled, or `"0"` / `"1"` / `"999"` once manually labeled
- Running `quickstart.py -s` again will **preserve existing labels** and only add newly discovered player:strategy pairs
